```python
import pandas as pd
import os
from sqlalchemy import create_engine
import logging
import time

# Setup logging
logging.basicConfig(
    filename="data_ingestion.log",
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode="a"
)

# Create SQLite engine
engine = create_engine("sqlite:///my_database.db", echo=False)


def ingest_data(df, table_name, engine):
    """Inject dataframe into database table safely"""
    # Replace invalid column names (e.g., spaces, keywords)
    df.columns = [c.strip().replace(" ", "_").replace("-", "_") for c in df.columns]
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",  # safer for first ingestion
        index=False,
        method="multi"
    )


def load_raw_data():
    """Load CSVs and ingest into database"""
    start = time.time()
    data_path = "data"
    
    for file in os.listdir(data_path):
        if file.endswith(".csv"):
            table_name = os.path.splitext(file)[0]  # remove .csv extension
            file_path = os.path.join(data_path, file)
            
            logging.info(f"Ingesting {file} into database (table: {table_name})")
            
            try:
                # Read CSV fully for small/medium files
                df = pd.read_csv(file_path)
                ingest_data(df, table_name, engine)
            except Exception as e:
                logging.error(f"Error ingesting {file}: {e}")
                print(f"Error ingesting {file}: {e}")
    
    end = time.time()
    total_time = (end - start) / 60
    logging.info("Ingestion complete")
    logging.info(f"Total time taken: {total_time:.2f} minutes")
    print(f"Data ingestion completed in {total_time:.2f} minutes.")


if __name__ == "__main__":
    load_raw_data()
```
